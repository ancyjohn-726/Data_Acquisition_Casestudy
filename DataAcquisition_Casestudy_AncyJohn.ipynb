Part B – Data Understanding

```python
#TASK 5

datasets = {
    'population': population_df,
    'gdp': gdp_df,
    'internet_users': internet_users_df,
    'literacy_rate': literacy_df
}

for name, df in datasets.items():
  print(f"\n{name}")
  print("rows and columns:", df.shape)
  print("Column names:",df.columns.tolist())
  print("\nData types:")
  print(df.dtypes)
  print("\nMissing values:")
  print(df.isnull().sum())
  print("Duplicate records:",df.duplicated().sum())
```

Part C – Data Transformation

```python
#TASK 6

population_df.rename(columns={'Country':'country'}, inplace=True)
gdp_df.rename(columns={'Country':'country'}, inplace=True)
internet_users_df.rename(columns={'Country':'country'}, inplace=True)
literacy_df.rename(columns={'name':'country'}, inplace=True)
print(population_df.columns)
print(gdp_df.columns)
print(internet_users_df.columns)
print(literacy_df.columns)
```

```python
#TASK 7

population_df['population'] = pd.to_numeric(population_df['population'])
gdp_df['GDP'] = pd.to_numeric(gdp_df['GDP'])
internet_users_df['internet_users'] = pd.to_numeric(internet_users_df['internet_users'])
literacy_df['literacy_rate'] = pd.to_numeric(literacy_df['literacy_rate'])

print("Polulation")
print(population_df.dtypes)
print('\n')
print("GDP")
print(gdp_df.dtypes)
print('\n')
print("Internet")
print(internet_users_df.dtypes)
print('\n')
print("Literacy")
print(literacy_df.dtypes)
```

```python
# Task 8

import pandas as pd

merged_df = pd.merge(population_df, gdp_df, on="country", how="inner")
merged_df = pd.merge(merged_df, internet_users_df, on="country", how="inner")
merged_df = pd.merge(merged_df, literacy_df, on="country", how="inner")

merged_df
```

```python
# Task 9

merged_df["Internet Penetration (%)"] = (
    merged_df["internet_users"] / merged_df["population"] * 100
).round(2)

merged_df["GDP Per Capita"] = (
    merged_df["GDP"] * 1_000_000 / merged_df["population"]
).round(2)

print(merged_df.head())
```

PART D

```python
# Task 10

import sqlite3

conn = sqlite3.connect("country_statistics.db")

merged_df.to_sql(
    "country_statistics",
    conn,
    if_exists="replace",
    index=False
)

conn.close()
```
